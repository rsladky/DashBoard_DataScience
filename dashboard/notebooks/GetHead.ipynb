```python
# Cell: afficher les head de tous les CSV dans ../data/raw (et ../data/row si présent)
import pandas as pd
from pathlib import Path
from IPython.display import display, Markdown

data_dirs = [Path("../data/raw"), Path("../data/row")]
csv_files = []
for d in data_dirs:
    if d.exists():
        csv_files.extend(sorted(d.glob("*.csv")))

if not csv_files:
    raise FileNotFoundError("Aucun fichier .csv trouvé dans ../data/raw ni ../data/row")

def read_head(path, n=5):
    for enc in ("utf-8", "latin-1", "cp1252"):
        try:
            return pd.read_csv(path, nrows=n, encoding=enc, engine="python")
        except Exception:
            continue
    # dernier recours
    return pd.read_csv(path, nrows=n, engine="python", encoding_errors="replace")

for p in csv_files:
    try:
        # essayer d'afficher un chemin relatif lisible
        try:
            rel = p.relative_to(Path.cwd())
        except Exception:
            try:
                rel = p.relative_to(Path.cwd().parent)
            except Exception:
                rel = p.name  # fallback
        display(Markdown(f"### {rel}"))
        df = read_head(p, n=10)
        display(df)
    except Exception as e:
        display(Markdown(f"**Erreur en lisant {p.name}:** {e}"))
```
